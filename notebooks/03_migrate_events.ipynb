```python
from timeline.storage.repository import SQLiteRepository
from timeline.api.timeline import Timeline
from timeline.domain.models import Event
from pathlib import Path
```

**Change event name/category based on name**

```python
old_category = "Home"
old_name = "Working"
new_category = "Mind"
new_name = "Working"
```

```python
with SQLiteRepository(Path("../timeline.db")) as repo:
    repo.ensure_schema()
    timeline = Timeline(repo)
    timeline = timeline.category(old_category).activity(old_name)
    events = timeline.events()
```

```python
for e in events:
    e.category = new_category
    e.name = new_name
    
```

```python
with SQLiteRepository(Path("../timeline.db")) as repo:
    for e in events:
        repo.update(e)
```
