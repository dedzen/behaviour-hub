Find and delete all events with specific source/category

```python
from timeline.storage.repository import SQLiteRepository
from timeline.api.timeline import Timeline
from timeline.domain.models import Event
from timeline.domain.enums import DeviceSource, EventKind
from pathlib import Path
```

```python
with SQLiteRepository(Path("../timeline.db")) as repo:
    repo.ensure_schema()
    timeline = Timeline(repo)
    timeline = timeline.source(DeviceSource.PHONE)
    events = timeline.events()
```

```python
events
```

```python
with SQLiteRepository(Path("../timeline.db")) as repo:
    for e in events:
        repo.delete(e, e.id) #type: ignore
```
